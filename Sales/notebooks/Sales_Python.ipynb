```python
# کتابخانه برای خواندن فایل
import csv
# باز کردن فایل
f = open(r"sales.csv", "r")
# خواندن فایل و تبدیل داده‌ها به دیکشنری
sales = list(csv.DictReader(f))
```

```python
# شمارش تعداد تراکنش‌ها
count = 0
# بررسی تمام ردیف‌های جدول
for row in sales:
    # بررسی اینکه کد تراکنش خالی نباشد
    if row["transaction_id"] != "":
        count += 1

print("Total transactions:", count)
```

```python
# مجموع فروش
total = 0
valid_count = 0

# بررسی تمام تراکنش‌ها
for row in sales:
    try:

        # ردیف بدون کد تراکنش نامعتبر است
        if row["transaction_id"] == "":
            print("Invalid Value: missing transaction_id")
            continue

        # تبدیل داده‌ها به عدد
        age = int(row["age"])
        quantity = int(row["quantity"])
        unit_price = float(row["unit_price"])
        discount = float(row["discount"])

        # بررسی معتبر بودن داده‌ها
        if not 0 < age <= 100 or quantity <= 0 or unit_price <= 0 or not 0 <= discount <= 100:
            print("Invalid Value: " , row["transaction_id"])
            continue

        # محاسبه مبلغ نهایی
        amount = (quantity * unit_price) * (1 - discount / 100)

        # اضافه کردن مبلغ به مجموع فروش
        total += amount
        valid_count += 1

    except (ValueError, TypeError):
        print("Invalid Value:", row["transaction_id"])
        continue

print("Valid transactions:", valid_count)
print("Total sales:", total)
```

```python
# محاسبه میانگین
average = total / valid_count

print("Average transaction amount:", average)
```
